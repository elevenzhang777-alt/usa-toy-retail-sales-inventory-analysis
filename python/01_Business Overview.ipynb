```python
# Import libraries for data analysis and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the daily performance dataset created in SQL
daily = pd.read_csv("daily_performance.csv")

# Convert Date to datetime format
daily["Date"] = pd.to_datetime(daily["Date"])

# Display basic information
print("Dataset shape:", daily.shape)
daily.head()
```

```python
# Check data types and missing values
daily.info()

print("\nMissing values:")
print(daily.isna().sum())
```

```python
# Calculate annual business KPIs from the daily analytical dataset
total_units = daily["Units_Sold"].sum()
total_revenue = daily["Revenue"].sum()
total_cogs = daily["COGS"].sum()
total_gross_profit = daily["Gross_Profit"].sum()
gross_margin = total_gross_profit / total_revenue

print(f"Total Units Sold: {total_units:,.0f}")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total COGS: ${total_cogs:,.2f}")
print(f"Total Gross Profit: ${total_gross_profit:,.2f}")
print(f"Gross Margin: {gross_margin:.2%}")
```

```python
# Plot daily revenue trend across 2025
plt.figure(figsize=(12, 5))
plt.plot(daily["Date"], daily["Revenue"])
plt.title("Daily Revenue Trend - 2025")
plt.xlabel("Date")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Plot daily revenue and gross profit trends
plt.figure(figsize=(12, 5))
plt.plot(daily["Date"], daily["Revenue"], label="Revenue")
plt.plot(daily["Date"], daily["Gross_Profit"], label="Gross Profit")
plt.title("Daily Revenue and Gross Profit Trend - 2025")
plt.xlabel("Date")
plt.ylabel("Amount ($)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Examine the distribution of daily gross margins
print(daily["Gross_Margin"].describe())

print(f"\nMinimum daily margin: {daily['Gross_Margin'].min():.2%}")
print(f"Maximum daily margin: {daily['Gross_Margin'].max():.2%}")
print(f"Average daily margin: {daily['Gross_Margin'].mean():.2%}")
```
